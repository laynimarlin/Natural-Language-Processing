```python
import pandas as pd

def analyze_pos(text, nlp_model):
    doc = nlp_model(text)
    results = []
    for token in doc:
        results.append({
            "Token": token.text,
            "POS": token.pos_,
            "POS Tag": token.tag_,
            "Lemma": token.lemma_,
            "Dependency": token.dep_,
            "Head": token.head.text
        })
    return pd.DataFrame(results)

df_pos = analyze_pos("Mahasiswa belajar NLP.", nlp)
print(df_pos.to_string(index=False))
```

```python
sentences = [
    "Mahasiswa belajar NLP.",
    "Dosen mengajar Natural Language Processing.",
    "Mahasiswa membaca buku.",
    "Sistem memproses data teks.",
    "Model bahasa memprediksi kata berikutnya."
]

for sentence in sentences:
    print(f"\nKalimat: {sentence}")
    df_result = analyze_pos(sentence, nlp)
    print(df_result[["Token", "POS", "Dependency", "Head"]].to_string(index=False))
    print()
```

```python
# Menerapkan POS tagging ke dataset:
import pandas as pd

df = pd.read_csv("scrape_kompas_clean.csv")
text_column = "title"  # sesuaikan

# Ambil sample untuk analisis
sample = df.sample(10, random_state=42)

for _, row in sample.iterrows():
    text = str(row[text_column])[:200]  # batasi panjang
    doc = nlp(text)
    pos_counts = {}
    for token in doc:
        pos = token.pos_
        pos_counts[pos] = pos_counts.get(pos, 0) + 1
    print(f"Teks: {text[:80]}...")
    print(f"POS Distribution: {pos_counts}")
    print()
```

```python
import spacy
import pandas as pd
from spacy import displacy

nlp = spacy.load("xx_ent_wiki_sm")
```

```python
def analisis_kalimat(teks):
    doc = nlp(teks)
    print(f"Kalimat: {teks}")
    print("-" * 70)
    
    # Simpan hasil ke tabel
    hasil = []
    for token in doc:
        hasil.append({
            "Token": token.text,
            "POS": token.pos_,
            "POS Tag": token.tag_,
            "Lemma": token.lemma_,
            "Dependency": token.dep_,
            "Head": token.head.text
        })
        print(f"{token.text:<20} {token.pos_:<8} {token.dep_:<15} → {token.head.text}")
    
    # Cek kesalahan
    kesalahan = []
    for token in doc:
        if token.pos_ == "X":
            kesalahan.append(f"Kata '{token.text}' tidak dikenali")
    if kesalahan:
        print(f"\n Kesalahan: {', '.join(kesalahan)}")
    else:
        print("\n Tidak ada kesalahan")
    
    # Tampilkan diagram dependensi
    displacy.render(doc, style="dep", jupyter=True, options={"distance": 90})
    
    print("\n")
    return pd.DataFrame(hasil)
```

```python
# Kalimat 1
df1 = analisis_kalimat("Pangsa Bank Syariah Masih 8 Persen, BSI Perluas Jangkauan ke Semua Segmen")

# Kalimat 2
df2 = analisis_kalimat("Prabowo Lantik Sudaryono Jadi Kepala BGN Gantikan Nanik S Deyang")

# Kalimat 3
df3 = analisis_kalimat("Bobotoh dan Aremania Satu Tribun, Sinyal Positif dari Piala Presiden 2026")

# Kalimat 4
df4 = analisis_kalimat("JKN Jadi Fondasi Perlindungan Kesehatan Anak Menuju Indonesia Emas 2045")

# Kalimat 5
df5 = analisis_kalimat("Partai Buruh Pilih Henry Saragih Jadi Sekjen Baru")

# Kalimat 6
df6 = analisis_kalimat("BGN Tanggung Biaya Korban Keracunan Susu MBG di Blora Sebelum Ditetapkan KLB")

# Kalimat 7
df7 = analisis_kalimat("Perluasan Peran Koperasi: Dari Sumur Minyak hingga PLTS")

# Kalimat 8
df8 = analisis_kalimat("Hak dan Kewajiban Ojol Akan Diatur dalam Undang-undang Tersendiri")

# Kalimat 9
df9 = analisis_kalimat("Potret Warga Medan Antre BBM di Tengah Banjir, Pengendara Pilih Bertahan Berjam-jam")

# Kalimat 10
df10 = analisis_kalimat("Inggris Nasionalisasi British Steel, Selamatkan Industri Baja dan 2.700 Pekerjaan")
```
