```python
import pandas as pd
import nltk
from nltk.tokenize import sent_tokenize, word_tokenize

nltk.download("punkt")
nltk.download("punkt_tab")
```

```python
df = pd.read_csv("scrape_kompas_clean.csv")
text_column = "title"  # sesuaikan dengan nama kolom
```

```python
#Memecah teks menjadi kalimat:
text = "Saya belajar NLP. NLP sangat menarik. Saya ingin menjadi NLP Engineer."

sentences = sent_tokenize(text)

for sentence in sentences:
    print(sentence)
```

```python
from nltk.tokenize import word_tokenize

text = "Saya belajar Natural Language Processing."

tokens = word_tokenize(text)

print(tokens)
```

```python
text = "NLP"

characters = list(text)

print(characters)
```

```python
#Menerapkan word tokenization ke seluruh dataset:
df["tokens"] = df[text_column].astype(str).apply(word_tokenize)

df[[text_column, "tokens"]].head()
```

```python
#Menghitung jumlah token per dokumen:
df["jumlah_token"] = df["tokens"].apply(len)
df["jumlah_token"].describe()
```

```python
#Hitung jumlah kata dengan split sederhana:
df["jumlah_kata"] = df[text_column].astype(str).apply(
    lambda x: len(x.split())
)

#Bandingkan
df[["jumlah_kata", "jumlah_token"]].describe()
```

```python
#Amati bagaimana tokenizer menangani berbagai jenis teks:
texts = [
    "Saya suka produk ini.",
    "Bagus banget!!!",
    "Tidak bagus...",
    "Harga Rp50.000.",
    "email@example.com",
    "https://example.com",
    "yg bagus bgt sih",
    "aku udah beli nih kak"
]

for text in texts:
    print("TEXT :", text)
    print("TOKEN:", word_tokenize(text))
    print()
```

```python
import pandas as pd
from nltk.tokenize import word_tokenize

# Teks yang akan diuji
test_texts = [
    "Saya belajar NLP.",
    "Bagus banget!!!",
    "Harga Rp50.000"
]

# Hasil tokenisasi
nltk_hasil = [word_tokenize(t) for t in test_texts]
split_hasil = [t.split() for t in test_texts]

# Tabel perbandingan
tabel = pd.DataFrame({
    "Teks": test_texts,
    "NLTK word_tokenize": nltk_hasil,
    "split()": split_hasil
})

tabel
```
